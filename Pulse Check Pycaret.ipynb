```python
# Import libraries

import pandas as pd
from pycaret.regression import *
```

```python
# Read cleaned Zillow data into a dataframe and preview

sdf = spark.table("workspace.zillow.cleaned_data")
display(sdf)
```

```python
sdf.count()
```

```python
display(sorted(sdf.columns))
```

```python
# Create pandas dataframe (required for pycaret) with subset of columns

required_columns = [
    'price',
    'bathrooms',
    'bedrooms',
    'city',
    # 'country',
    'homestatus',
    'hometype',
    'state',
    # 'street',
    'yearbuilt',
    'zipcode'
]

df = sdf.toPandas()[required_columns]
```

```python
display(df.info())
```

```python
# Convert numerical columns to proper datatype

numeric_columns = [
    'price',
    'bathrooms',
    'bedrooms',
    'zipcode',
    'yearbuilt'
]
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.info()
```

```python
# Start a Pycaret experiment
exp = setup(
    df,
    target='price', # target variable to predict
    train_size=0.8, # % of dataset to use as training
    memory='/Volumes/workspace/zillow/temp'  # specify a directory with write permissions
)
```

```python
# Compare regression models

best_model = compare_models(
    sort='R2', # sort model metrics by which column
    round=2, # round model metrics
    n_select=1 # number of models to create
)
```

```python
plot_model(best_model, plot='residuals')
```

```python
plot_model(best_model, plot='feature')
```
